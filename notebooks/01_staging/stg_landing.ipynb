# Staging: Landing

## Setup

### Imports

```python
from pathlib import Path

import duckdb
```

### Configuration

```python
# Locate the project root (folder containing requirements.txt), independent of cwd
PROJECT_ROOT = next(
    p for p in [Path.cwd(), *Path.cwd().parents] if (p / "requirements.txt").exists()
)

data_vault_path = str(PROJECT_ROOT / "data" / "warehouse" / "data_vault.duckdb")
```

```python
data_folder = PROJECT_ROOT / "data" / "raw"
```

### DuckDB Connection

```python
con = duckdb.connect(data_vault_path)
```

```python
con.execute("""
    CREATE SCHEMA IF NOT EXISTS staging;
""")
```

## Load Data

```python
for file_path in data_folder.glob("*.csv"):
    table_name = file_path.stem.replace("_dataset", "").replace("olist_", "")

    con.execute(f"""
        CREATE OR REPLACE TABLE staging.raw_{table_name} AS
        SELECT *
        FROM read_csv_auto('{file_path.as_posix()}');
    """)

    print(f"Loaded {file_path.name} -> staging.raw_{table_name}")
```
