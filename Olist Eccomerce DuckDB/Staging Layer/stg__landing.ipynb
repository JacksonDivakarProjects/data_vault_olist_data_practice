```python
data_folder = Path(r"C:\Users\JacksonDivakarRajasi\OneDrive - JMAN Group Ltd\Documents\Python Folder\DuckDB Environment\Olist Ecommerce Data Vault Dataset")
```

#### Connect to DuckDB

```python
con = duckdb.connect(data_vault_path)
```

```python
con.execute("""
    CREATE SCHEMA IF NOT EXISTS staging;
""")
```

#### Load Data

```python
for file_path in data_folder.glob("*.csv"):
    table_name = file_path.stem.replace("_dataset", "").replace("olist_", "")

    con.execute(f"""
        CREATE OR REPLACE TABLE staging.raw_{table_name} AS
        SELECT *
        FROM read_csv_auto('{file_path.as_posix()}');
    """)

    print(f"Loaded {file_path.name} -> staging.raw_{table_name}")
```
